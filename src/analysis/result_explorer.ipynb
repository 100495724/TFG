# Act 1 — Result Explorer

Notebook organizado por basket. Cada basket (A1–A15) tiene cuatro celdas:

1. Descripción del archetype
2. Gráficas de evolución de allocation (control / gender / geo)
3. Tablas de decisión (action / allocation / reasoning)
4. Tablas de métricas (agent gap / committee gap / dynamic summary)

Para inspeccionar un basket concreto, ejecuta solo sus celdas tras la celda de **Setup**.

## Setup

```python
%matplotlib inline

from result_explorer import (
    load_result_csvs,
    plot_pair_all_variants,
    display_pair_tables_all_fields,
    display_pair_metrics_for_basket,
    available_pairs,
    available_seeds,
)

# ----- Configuración del análisis -----
RESULTS_PATHS = ["../results/act1_baseline_seed42.csv"]   # ajusta o añade más paths/globs
SEED = 42
INCLUDE_ERRORS = False

df = load_result_csvs(RESULTS_PATHS)

# Fix de variantes (alineamos variant con sensitive_attr)
df.loc[df["sensitive_attr"] == "gender",  "variant"] = "gender"
df.loc[df["sensitive_attr"] == "country", "variant"] = "geo"

print("Available pairs:", available_pairs(df))
print("Available seeds:", available_seeds(df))
```

## Basket A1 — HighGrowth_HighDebt

**Archetype**: Alto crecimiento financiado con deuda elevada.

Tres variantes: `control` (sin atributo sensible), `gender` (variación en género del CEO), `geo` (variación en origen geográfico).

### A1 · Gráficas de evolución

```python
# Plots de allocation por turno para las 3 variantes de A1
_ = plot_pair_all_variants(
    df,
    pair_id="A1",
    seed=SEED,
    output_dir=None,    # None → no guarda, muestra inline
    include_errors=INCLUDE_ERRORS,
    show=True,
)
```

### A1 · Tablas de decisión

```python
# Tablas pivote (turno × agente) para action / allocation / reasoning
_ = display_pair_tables_all_fields(
    df,
    pair_id="A1",
    seed=SEED,
    fields=("action", "allocation", "reasoning"),
    include_errors=INCLUDE_ERRORS,
)
```

### A1 · Tablas de métricas

```python
# Métricas filtradas a este basket: agent gap / committee gap / dynamic summary
_ = display_pair_metrics_for_basket(
    df,
    pair_id="A1",
    seed=SEED,
    include_errors=INCLUDE_ERRORS,
)
```
